```python
from pathlib import Path

# ============================================================
# CONFIGURACIÓN
# ============================================================

WHITE_DIR = Path("/content/test_white")
COMPLEX_DIR = Path("/content/test_complex")

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}


# ============================================================
# BUSCAR CARPETAS images Y labels
# ============================================================

def find_folder(root, folder_name):
    folders = [
        p for p in Path(root).rglob(folder_name)
        if p.is_dir()
    ]

    if len(folders) == 0:
        raise FileNotFoundError(
            f"❌ No encontré una carpeta '{folder_name}' dentro de {root}"
        )

    return folders[0]


white_images = find_folder(WHITE_DIR, "images")
white_labels = find_folder(WHITE_DIR, "labels")

complex_images = find_folder(COMPLEX_DIR, "images")
complex_labels = find_folder(COMPLEX_DIR, "labels")


# ============================================================
# MOSTRAR RUTAS DETECTADAS
# ============================================================

print("RUTAS DETECTADAS\n")

print("WHITE:")
print(" Images:", white_images)
print(" Labels:", white_labels)

print("\nCOMPLEX:")
print(" Images:", complex_images)
print(" Labels:", complex_labels)


# ============================================================
# FUNCIÓN DE AUDITORÍA
# ============================================================

def audit_subset(name, images_dir, labels_dir, expected):

    images = sorted([
        p for p in images_dir.iterdir()
        if p.suffix.lower() in IMAGE_EXTENSIONS
    ])

    labels = sorted(labels_dir.glob("*.txt"))

    image_stems = {p.stem for p in images}
    label_stems = {p.stem for p in labels}

    missing_labels = image_stems - label_stems
    orphan_labels = label_stems - image_stems

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print(f"Imágenes encontradas : {len(images)}")
    print(f"Labels encontrados   : {len(labels)}")
    print(f"Imágenes esperadas   : {expected}")
    print(f"Imágenes sin label   : {len(missing_labels)}")
    print(f"Labels sin imagen    : {len(orphan_labels)}")

    if (
        len(images) == expected
        and len(labels) == expected
        and len(missing_labels) == 0
        and len(orphan_labels) == 0
    ):
        print("\n✅ SUBSET CORRECTO")
    else:
        print("\n⚠️ REVISAR SUBSET")

    if missing_labels:
        print("\n❌ Imágenes sin label:")
        for x in sorted(missing_labels):
            print(" -", x)

    if orphan_labels:
        print("\n❌ Labels sin imagen:")
        for x in sorted(orphan_labels):
            print(" -", x)

    return images, labels


# ============================================================
# AUDITAR WHITE
# ============================================================

white_imgs, white_lbls = audit_subset(
    "WHITE BACKGROUND",
    white_images,
    white_labels,
    103
)


# ============================================================
# AUDITAR COMPLEX
# ============================================================

complex_imgs, complex_lbls = audit_subset(
    "COMPLEX SCENE",
    complex_images,
    complex_labels,
    91
)


# ============================================================
# RESULTADO FINAL
# ============================================================

if len(white_imgs) == 103 and len(complex_imgs) == 91:
    print("\n" + "=" * 70)
    print("🎯 TOTAL TEST:", len(white_imgs) + len(complex_imgs))
    print("White background:", len(white_imgs))
    print("Complex scene:", len(complex_imgs))
    print("=" * 70)
```

```python
import pandas as pd

def count_objects(scene, images, labels_dir):
    rows = []

    for img in images:
        label_file = labels_dir / f"{img.stem}.txt"

        with open(label_file, "r") as f:
            annotations = [
                line.strip()
                for line in f
                if line.strip()
            ]

        rows.append({
            "image_name": img.name,
            "scene_type": scene,
            "objects": len(annotations)
        })

    return rows


rows = []

rows += count_objects(
    "White background",
    white_imgs,
    white_labels
)

rows += count_objects(
    "Complex scene",
    complex_imgs,
    complex_labels
)

objects_df = pd.DataFrame(rows)

print("=" * 70)
print("OBJECTS PER IMAGE")
print("=" * 70)

summary = (
    objects_df
    .groupby("scene_type")["objects"]
    .agg(["count", "sum", "mean", "std", "min", "max"])
    .round(2)
)

display(summary)

print("\nTOTAL:")
print("Images :", len(objects_df))
print("Objects:", objects_df["objects"].sum())

objects_df.to_csv(
    "/content/objects_per_image.csv",
    index=False
)

print("\n✅ Guardado: /content/objects_per_image.csv")
```

```python
# ============================================================
# R1 - 30 CHECKPOINTS × 2 SCENES = 60 EVALUATIONS
# ============================================================

import sys
import re
import gc
import pandas as pd
import torch
from pathlib import Path

from ultralytics import YOLO

# YOLOv5
sys.path.insert(0, "/content/yolov5")
import val as yolov5_val


# ============================================================
# CONFIG
# ============================================================

CHECKPOINT_DIR = Path("/content/ENTRENAMIENTOS_PT_VF")

SCENES = {
    "white_background": "/content/white.yaml",
    "complex_scene": "/content/complex.yaml"
}

OUTPUT_CSV = Path("/content/R1_ALL_RUNS.csv")


# ============================================================
# CHECKPOINTS
# ============================================================

checkpoints = sorted(CHECKPOINT_DIR.rglob("best_*.pt"))

print(f"Checkpoints encontrados: {len(checkpoints)}")

if len(checkpoints) != 30:
    raise ValueError(
        f"❌ Se esperaban 30 checkpoints, pero se encontraron {len(checkpoints)}"
    )


# ============================================================
# EXTRAER MODELO Y RUN DEL NOMBRE
# best_v5n_1.pt -> v5n / run 1
# ============================================================

def parse_checkpoint(path):

    match = re.match(
        r"best_(v[58][nsm])_(\d+)\.pt",
        path.name.lower()
    )

    if not match:
        raise ValueError(
            f"Nombre de checkpoint no reconocido: {path.name}"
        )

    model = match.group(1)
    run = int(match.group(2))

    return model, run


# ============================================================
# CARGAR RESULTADOS EXISTENTES
# Permite reanudar si Colab falla
# ============================================================

if OUTPUT_CSV.exists():

    results_df = pd.read_csv(OUTPUT_CSV)

    print(
        f"🔄 Archivo existente encontrado: "
        f"{len(results_df)} evaluaciones guardadas."
    )

else:

    results_df = pd.DataFrame(columns=[
        "model",
        "run",
        "scene",
        "precision",
        "recall",
        "mAP50",
        "mAP50_95"
    ])


# ============================================================
# VERIFICAR SI UNA EVALUACIÓN YA EXISTE
# ============================================================

def already_done(model, run, scene):

    if len(results_df) == 0:
        return False

    condition = (
        (results_df["model"] == model) &
        (results_df["run"] == run) &
        (results_df["scene"] == scene)
    )

    return condition.any()


# ============================================================
# LOOP PRINCIPAL
# ============================================================

total_expected = 60
counter = len(results_df)

for checkpoint in checkpoints:

    model_name, run = parse_checkpoint(checkpoint)

    for scene_name, yaml_path in SCENES.items():

        # ----------------------------------------------------
        # Saltar evaluación ya realizada
        # ----------------------------------------------------

        if already_done(model_name, run, scene_name):

            print(
                f"\n⏭️ YA EXISTE: "
                f"{model_name.upper()} Run {run} / {scene_name}"
            )

            continue


        counter += 1

        print("\n" + "=" * 75)

        print(
            f"[{counter}/{total_expected}] "
            f"{model_name.upper()} "
            f"RUN {run} "
            f"/ {scene_name.upper()}"
        )

        print("=" * 75)


        # ====================================================
        # YOLOv5
        # ====================================================

        if model_name.startswith("v5"):

            results, maps, times = yolov5_val.run(

                data=yaml_path,
                weights=str(checkpoint),

                batch_size=16,
                imgsz=640,

                task="test",
                device="0",

                plots=False,
                verbose=False,

                project="/content/runs/R1_YOLOv5",

                name=f"{model_name}_run{run}_{scene_name}",

                exist_ok=True
            )

            precision = float(results[0])
            recall = float(results[1])
            map50 = float(results[2])
            map5095 = float(results[3])


        # ====================================================
        # YOLOv8
        # ====================================================

        elif model_name.startswith("v8"):

            yolo = YOLO(str(checkpoint))

            metrics = yolo.val(

                data=yaml_path,
                split="test",

                imgsz=640,
                batch=16,

                device=0,

                plots=False,
                verbose=False,

                project="/content/runs/R1_YOLOv8",

                name=f"{model_name}_run{run}_{scene_name}"
            )

            precision = float(metrics.box.mp)
            recall = float(metrics.box.mr)
            map50 = float(metrics.box.map50)
            map5095 = float(metrics.box.map)


        # ====================================================
        # GUARDAR
        # ====================================================

        new_row = pd.DataFrame([{

            "model": model_name,
            "run": run,
            "scene": scene_name,

            "precision": precision,
            "recall": recall,

            "mAP50": map50,
            "mAP50_95": map5095

        }])

        results_df = pd.concat(
            [results_df, new_row],
            ignore_index=True
        )

        # Guardar INMEDIATAMENTE
        results_df.to_csv(
            OUTPUT_CSV,
            index=False
        )


        print("\nRESULTADO:")

        print(f"Precision : {precision:.4f}")
        print(f"Recall    : {recall:.4f}")
        print(f"mAP50     : {map50:.4f}")
        print(f"mAP50-95  : {map5095:.4f}")

        print(
            f"\n💾 Guardado: "
            f"{len(results_df)}/{total_expected}"
        )


        # Liberar memoria
        if model_name.startswith("v8"):
            del yolo
            del metrics

        gc.collect()
        torch.cuda.empty_cache()


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 75)
print("🎯 R1 TERMINADO")
print("=" * 75)

print(
    f"Evaluaciones guardadas: "
    f"{len(results_df)}/60"
)

print(
    f"Archivo: {OUTPUT_CSV}"
)

display(
    results_df.sort_values(
        ["model", "run", "scene"]
    )
)
```

```python
# ============================================================
# R1 - RESUMEN FINAL
# ============================================================

import pandas as pd

df = pd.read_csv("/content/R1_ALL_RUNS.csv")


# ------------------------------------------------------------
# 1. MEAN ± SD POR MODELO Y ESCENA
# ------------------------------------------------------------

summary = (
    df
    .groupby(["model", "scene"])
    .agg(
        n=("mAP50_95", "count"),

        precision_mean=("precision", "mean"),
        precision_sd=("precision", "std"),

        recall_mean=("recall", "mean"),
        recall_sd=("recall", "std"),

        mAP50_mean=("mAP50", "mean"),
        mAP50_sd=("mAP50", "std"),

        mAP50_95_mean=("mAP50_95", "mean"),
        mAP50_95_sd=("mAP50_95", "std")
    )
    .reset_index()
)

summary.to_csv(
    "/content/R1_MEAN_SD.csv",
    index=False
)

print("MEAN ± SD")
display(summary.round(4))


# ------------------------------------------------------------
# 2. Δscene POR CADA EJECUCIÓN
#
# Complex - White
# ------------------------------------------------------------

paired = df.pivot(
    index=["model", "run"],
    columns="scene",
    values="mAP50_95"
).reset_index()

paired["delta_scene"] = (
    paired["complex_scene"]
    -
    paired["white_background"]
)

print("\nΔscene POR RUN")
display(paired.round(4))


# ------------------------------------------------------------
# 3. RESUMEN Δscene
# ------------------------------------------------------------

delta_summary = (
    paired
    .groupby("model")
    .agg(
        delta_mean=("delta_scene", "mean"),
        delta_sd=("delta_scene", "std"),
        delta_min=("delta_scene", "min"),
        delta_max=("delta_scene", "max")
    )
    .reset_index()
)

delta_summary.to_csv(
    "/content/R1_SCENE_DELTA.csv",
    index=False
)

print("\nΔscene RESUMEN")
display(delta_summary.round(4))
```
