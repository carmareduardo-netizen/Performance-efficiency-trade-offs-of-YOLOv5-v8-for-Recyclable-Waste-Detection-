```python
from pathlib import Path

def contar_imagenes(folder):
    extensiones = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

    return sum(
        1
        for archivo in Path(folder).iterdir()
        if archivo.is_file()
        and archivo.suffix.lower() in extensiones
    )


TRAIN_IMAGES = "/content/DATASET/train/images"
VALID_IMAGES = "/content/DATASET/valid/images"
TEST_IMAGES  = "/content/DATASET/test/images"


n_train = contar_imagenes(TRAIN_IMAGES)
n_valid = contar_imagenes(VALID_IMAGES)
n_test  = contar_imagenes(TEST_IMAGES)

total = n_train + n_valid + n_test


print("=" * 40)
print("DATASET")
print("=" * 40)

print(f"TRAIN : {n_train}")
print(f"VALID : {n_valid}")
print(f"TEST  : {n_test}")

print("-" * 40)

print(f"TOTAL : {total}")
```

```python
from ultralytics import YOLO
import pandas as pd
import numpy as np
import os
import re
import gc
import torch
from pathlib import Path

# ============================================================
# 1. LOCALIZAR LOS 15 PESOS YOLOv8
# ============================================================

v8_weights = sorted([
    w for w in weights
    if "best_v8" in os.path.basename(w).lower()
])

print("Pesos YOLOv8 encontrados:", len(v8_weights))

if len(v8_weights) != 15:
    print("⚠️ ADVERTENCIA: se esperaban 15 pesos YOLOv8.")

for w in v8_weights:
    print(os.path.basename(w))


# ============================================================
# 2. FUNCIÓN PARA IDENTIFICAR MODELO Y RUN
# ============================================================

def get_model_run(weight_path):

    filename = os.path.basename(weight_path)

    # Ejemplo: best_v8m_1.pt
    match = re.search(
        r"best_v8([nsm])_(\d+)",
        filename.lower()
    )

    if match:
        scale = match.group(1)
        run = int(match.group(2))

        model = f"YOLOv8{scale}"

        return model, run

    return "UNKNOWN", 0


# ============================================================
# 3. EVALUAR LOS 15 PESOS
# ============================================================

results_list = []

for i, weight_path in enumerate(v8_weights, start=1):

    model_name, run = get_model_run(weight_path)

    print("\n" + "=" * 65)
    print(
        f"[{i}/{len(v8_weights)}] "
        f"{model_name} - Run {run}"
    )
    print("=" * 65)

    try:

        # ---------------------------------------------
        # CARGAR MODELO
        # ---------------------------------------------

        model = YOLO(weight_path)

        # ---------------------------------------------
        # TEST ORIGINAL
        # ---------------------------------------------

        print("Evaluando TEST ORIGINAL (194 imágenes)...")

        original = model.val(
            data="/content/test_original.yaml",
            split="test",
            imgsz=640,
            batch=16,
            plots=False,
            verbose=False
        )

        original_p = float(original.box.mp)
        original_r = float(original.box.mr)
        original_map50 = float(original.box.map50)
        original_map5095 = float(original.box.map)

        # ---------------------------------------------
        # TEST CLEAN
        # ---------------------------------------------

        print("Evaluando TEST CLEAN (175 imágenes)...")

        clean = model.val(
            data="/content/test_clean.yaml",
            split="test",
            imgsz=640,
            batch=16,
            plots=False,
            verbose=False
        )

        clean_p = float(clean.box.mp)
        clean_r = float(clean.box.mr)
        clean_map50 = float(clean.box.map50)
        clean_map5095 = float(clean.box.map)

        # ---------------------------------------------
        # DELTAS
        # Clean - Original
        # ---------------------------------------------

        delta_p = clean_p - original_p
        delta_r = clean_r - original_r
        delta_map50 = clean_map50 - original_map50
        delta_map5095 = clean_map5095 - original_map5095

        # ---------------------------------------------
        # GUARDAR RESULTADO
        # ---------------------------------------------

        results_list.append({

            "Model": model_name,
            "Run": run,
            "Weight": os.path.basename(weight_path),

            "Original_Precision": original_p,
            "Clean_Precision": clean_p,
            "Delta_Precision": delta_p,

            "Original_Recall": original_r,
            "Clean_Recall": clean_r,
            "Delta_Recall": delta_r,

            "Original_mAP50": original_map50,
            "Clean_mAP50": clean_map50,
            "Delta_mAP50": delta_map50,

            "Original_mAP50_95": original_map5095,
            "Clean_mAP50_95": clean_map5095,
            "Delta_mAP50_95": delta_map5095
        })

        # ---------------------------------------------
        # MOSTRAR RESULTADO DE ESTA CORRIDA
        # ---------------------------------------------

        print("\nRESULTADO:")

        print(
            f"Precision: "
            f"{original_p:.4f} -> {clean_p:.4f} "
            f"(Δ {delta_p:+.4f})"
        )

        print(
            f"Recall: "
            f"{original_r:.4f} -> {clean_r:.4f} "
            f"(Δ {delta_r:+.4f})"
        )

        print(
            f"mAP50: "
            f"{original_map50:.4f} -> {clean_map50:.4f} "
            f"(Δ {delta_map50:+.4f})"
        )

        print(
            f"mAP50-95: "
            f"{original_map5095:.4f} -> {clean_map5095:.4f} "
            f"(Δ {delta_map5095:+.4f})"
        )

        # Liberar memoria
        del model
        del original
        del clean

        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    except Exception as e:

        print("\n❌ ERROR con:", os.path.basename(weight_path))
        print(e)


# ============================================================
# 4. DATAFRAME CON LAS 15 CORRIDAS
# ============================================================

df = pd.DataFrame(results_list)

df = df.sort_values(
    ["Model", "Run"]
).reset_index(drop=True)

print("\n\n========================================")
print("RESULTADOS COMPLETOS")
print("========================================")

display(df)
```

```python
import pandas as pd
from google.colab import files

# Nombre del archivo
OUTPUT_EXCEL = "/content/YOLOv8_Leakage_Sensitivity.xlsx"

# Generar resumen por modelo
summary = df.groupby("Model").agg({
    "Original_Precision": "mean",
    "Clean_Precision": "mean",
    "Delta_Precision": "mean",
    "Original_Recall": "mean",
    "Clean_Recall": "mean",
    "Delta_Recall": "mean",
    "Original_mAP50": "mean",
    "Clean_mAP50": "mean",
    "Delta_mAP50": "mean",
    "Original_mAP50_95": "mean",
    "Clean_mAP50_95": "mean",
    "Delta_mAP50_95": "mean"
}).reset_index()

# Exportar
with pd.ExcelWriter(OUTPUT_EXCEL, engine="openpyxl") as writer:

    # Resultados de las 15 corridas
    df.to_excel(
        writer,
        sheet_name="15_RUNS",
        index=False
    )

    # Resumen por modelo
    summary.to_excel(
        writer,
        sheet_name="MODEL_SUMMARY",
        index=False
    )

print("Excel generado correctamente:")
print(OUTPUT_EXCEL)

# Descargar automáticamente
files.download(OUTPUT_EXCEL)
```

```python
import hashlib
import os

def sha256_file(filepath):
    sha = hashlib.sha256()
    with open(filepath, "rb") as f:
        while True:
            chunk = f.read(1024 * 1024)
            if not chunk:
                break
            sha.update(chunk)
    return sha.hexdigest()

print("=" * 100)
print("CHECK DE PESOS YOLOv5")
print("=" * 100)

for w in v5_weights:
    print(os.path.basename(w), " | ", sha256_file(w))
```

```python
import sys
import os
import pandas as pd

# Importar val.py de YOLOv5
sys.path.insert(0, "/content/yolov5")
import val

results = []

print("=" * 100)
print("EVALUACIÓN YOLOv5 - ORIGINAL vs CLEAN")
print("=" * 100)

for i, weight in enumerate(v5_weights, 1):

    filename = os.path.basename(weight)

    # Extraer modelo y run
    # Ejemplo: best_v5m_3.pt
    parts = filename.replace(".pt", "").split("_")
    model = parts[1].upper()   # V5M
    run = int(parts[2])

    print(f"\n[{i}/{len(v5_weights)}] {filename}")

    # ============================================================
    # TEST ORIGINAL
    # ============================================================

    print("  → Original (194 imágenes)")

    original_results, _, _ = val.run(
        data="/content/test_original.yaml",
        weights=weight,
        batch_size=16,
        imgsz=640,
        task="test",
        plots=False,
        verbose=False,
        project="/content/V5_BATCH",
        name=f"{model}_run{run}_original",
        exist_ok=True
    )

    P_o = original_results[0]
    R_o = original_results[1]
    mAP50_o = original_results[2]
    mAP5095_o = original_results[3]

    # ============================================================
    # TEST CLEAN
    # ============================================================

    print("  → Clean (175 imágenes)")

    clean_results, _, _ = val.run(
        data="/content/test_clean.yaml",
        weights=weight,
        batch_size=16,
        imgsz=640,
        task="test",
        plots=False,
        verbose=False,
        project="/content/V5_BATCH",
        name=f"{model}_run{run}_clean",
        exist_ok=True
    )

    P_c = clean_results[0]
    R_c = clean_results[1]
    mAP50_c = clean_results[2]
    mAP5095_c = clean_results[3]

    # ============================================================
    # GUARDAR RESULTADOS
    # ============================================================

    results.append({
        "Model": model,
        "Run": run,

        "Original_P": P_o,
        "Clean_P": P_c,
        "Delta_P": P_c - P_o,

        "Original_R": R_o,
        "Clean_R": R_c,
        "Delta_R": R_c - R_o,

        "Original_mAP50": mAP50_o,
        "Clean_mAP50": mAP50_c,
        "Delta_mAP50": mAP50_c - mAP50_o,

        "Original_mAP50-95": mAP5095_o,
        "Clean_mAP50-95": mAP5095_c,
        "Delta_mAP50-95": mAP5095_c - mAP5095_o
    })

    print(
        f"  ΔmAP50 = {mAP50_c-mAP50_o:+.4f} | "
        f"ΔmAP50-95 = {mAP5095_c-mAP5095_o:+.4f}"
    )

# ============================================================
# DATAFRAME
# ============================================================

df_v5 = pd.DataFrame(results)

df_v5 = df_v5.sort_values(
    ["Model", "Run"]
).reset_index(drop=True)

print("\n" + "=" * 100)
print("RESULTADOS")
print("=" * 100)

display(df_v5.round(6))
```

```python
summary_v5 = df_v5.groupby("Model").agg(
    Original_P_mean=("Original_P", "mean"),
    Clean_P_mean=("Clean_P", "mean"),
    Delta_P_mean=("Delta_P", "mean"),

    Original_R_mean=("Original_R", "mean"),
    Clean_R_mean=("Clean_R", "mean"),
    Delta_R_mean=("Delta_R", "mean"),

    Original_mAP50_mean=("Original_mAP50", "mean"),
    Clean_mAP50_mean=("Clean_mAP50", "mean"),
    Delta_mAP50_mean=("Delta_mAP50", "mean"),

    Original_mAP5095_mean=("Original_mAP50-95", "mean"),
    Clean_mAP5095_mean=("Clean_mAP50-95", "mean"),
    Delta_mAP5095_mean=("Delta_mAP50-95", "mean"),

    Delta_mAP50_std=("Delta_mAP50", "std"),
    Delta_mAP5095_std=("Delta_mAP50-95", "std")
).reset_index()

display(summary_v5.round(6))
```
